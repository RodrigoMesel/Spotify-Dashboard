##Top 10 Artistas e genêros mais escutados

```python
groupedArtists = df.groupby('artist_name')['stream_count'].sum()
top10Artists = groupedArtists.sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top10Artists.values, y=top10Artists.index, palette='mako')
plt.title('Top 10 Artistas com Mais Streams (2015-2025)')
plt.xlabel('Total de Streams')
plt.ylabel('Artista')
plt.show()
```

```python
groupedGenres = df.groupby('genre')['stream_count'].sum()
top10Genres = groupedGenres.sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top10Genres.values, y=top10Genres.index, palette='rocket')
plt.title('Top 10 Gêneros Musicais por Volume de Streams')
plt.xlabel('Total de Streams')
plt.ylabel('Gênero')
plt.show()
```

## Analise de quantidade de streams e popularidade ao longo dos anos

```python
streams_by_year = df.groupby('release_year')['stream_count'].sum().reset_index()

plt.figure(figsize=(10, 5))
sns.lineplot(data=streams_by_year, x='release_year', y='stream_count', marker='o', linewidth=2.5)
plt.title('Evolução do Volume de Streams por Ano (2015-2025)')
plt.xlabel('Ano')
plt.ylabel('Total de Streams')
plt.xticks(range(2015, 2026))
plt.show()
```

```python
pop_year = df.groupby('release_year')['popularity'].mean().reset_index()

plt.figure(figsize=(10, 5))
sns.lineplot(data=pop_year, x='release_year', y='popularity', marker='s', color='orange', linewidth=2.5)
plt.title('Popularidade Média das Músicas por Ano')
plt.xlabel('Ano')
plt.ylabel('Popularidade Média')
plt.xticks(range(2015, 2026))
plt.show()
```

## Porcentagem dos artistas responsáveis por 80% das streams

```python
artistas_streams = df.groupby('artist_name')['stream_count'].sum().sort_values(ascending=False)

streams_acumulados = artistas_streams.cumsum()
porcentagem_acumulada = streams_acumulados / artistas_streams.sum() * 100

top_80_percent = porcentagem_acumulada[porcentagem_acumulada <= 80]
proporcao_artistas_80 = (len(top_80_percent) / len(artistas_streams)) * 100

print(f"{proporcao_artistas_80:.2f}% dos artistas são responsáveis por 80% de todos os streams.")

plt.figure(figsize=(10, 6))
plt.plot(porcentagem_acumulada.values, color='teal', linewidth=2.5)
plt.title('Distribuição de Streams: A Cauda Longa dos Artistas')
plt.xlabel('Número de Artistas (Ordenados do maior para o menor)')
plt.ylabel('Porcentagem Acumulada de Streams (%)')
plt.legend()
plt.show()
```

## Ciclos de Tendência por Gênero

```python
genre_year_streams = df.groupby(['release_year', 'genre'])['stream_count'].sum().unstack().fillna(0)

genre_market_share = genre_year_streams.div(genre_year_streams.sum(axis=1), axis=0) * 100

top_5_genres_all_time = df.groupby('genre')['stream_count'].sum().nlargest(5).index
genre_market_share_top = genre_market_share[top_5_genres_all_time]

genre_market_share_top.plot(kind='area', stacked=True, figsize=(12, 7), colormap='Set2', alpha=0.8)
plt.title('Evolução do Market Share (Fatia de Streams) dos 5 Principais Gêneros (2015-2025)')
plt.xlabel('Ano')
plt.ylabel('Porcentagem do Total de Streams (%)')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```

## Analise One-Hit Wonders X Artistas consistentes

- O CV (Coeficiente de Variação) mede o quanto as músicas de um mesmo artista são diferentes entre si em número de plays.
- Artista Consistente (CV Baixo): Lança 10 músicas e todas batem ali na faixa de 50 milhões de streams. O público escuta a discografia inteira.
- One-Hit Wonder (CV Alto): Lança 10 músicas. Uma viraliza e bate 500 milhões de streams, mas as outras 9 não passam de 10 mil plays. O público só quer ouvir aquele único hit.

```python
artistas_contagem = df['artist_name'].value_counts()
artistas_frequentes = artistas_contagem[artistas_contagem >= 5].index
df_freq = df[df['artist_name'].isin(artistas_frequentes)]

consistencia = df_freq.groupby('artist_name')['stream_count'].agg(['mean', 'std', 'max']).dropna()
consistencia['CV'] = consistencia['std'] / consistencia['mean']

one_hit_wonders = consistencia.sort_values(by='CV', ascending=False).head(5)
consistentes = consistencia.sort_values(by=['CV', 'mean'], ascending=[True, False]).head(5)

plt.figure(figsize=(12, 7))
sns.scatterplot(data=consistencia, x='mean', y='CV', alpha=0.4, color='gray')

plt.scatter(one_hit_wonders['mean'], one_hit_wonders['CV'], color='red', s=100, label='One-Hit Wonders', edgecolor='black')
plt.scatter(consistentes['mean'], consistentes['CV'], color='green', s=100, label='Artistas Consistentes', edgecolor='black')

plt.title('Mapa de Consistência: Média de Streams vs. Coeficiente de Variação (CV)', fontsize=14)
plt.xlabel('Média de Streams por Faixa (Log Scale)', fontsize=12)
plt.ylabel('Variação entre as músicas (CV)', fontsize=12)
plt.xscale('log')
plt.legend()
plt.show()
```
